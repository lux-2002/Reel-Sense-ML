# ReelSense — Dataset Exploration

This notebook audits the complete Rotten Tomatoes movie dataset before modelling. The goal is to understand schema, missingness, duplicates and the target distribution without making modelling claims yet.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

DATA_PATH = Path('../data/Rotten_Tomatoes_Movies3.csv')
df = pd.read_csv(DATA_PATH)
print(df.shape)
df.head()
```

## Schema and data quality

```python
display(df.info())
display(df.isna().sum().sort_values(ascending=False).head(20))
print('Duplicate rows:', df.duplicated().sum())
print('Target missing:', df['audience_rating'].isna().sum())
```

## Target distribution

```python
ax = df['audience_rating'].dropna().plot(kind='hist', bins=20, figsize=(9,5), title='Audience Rating Distribution')
ax.set_xlabel('Audience rating')
plt.show()
```
